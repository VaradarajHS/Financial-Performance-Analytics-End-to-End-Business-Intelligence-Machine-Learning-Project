```python
import pandas as pd
import numpy as np
df = pd.read_csv("Financialdata_Cleaned.csv")
if {'Revenue', 'Gross_Profit'}.issubset(df.columns):
    df['Gross_Margin'] = (df['Gross_Profit'] / df['Revenue']) * 100

if {'Revenue', 'Operating_Income'}.issubset(df.columns):
    df['Operating_Margin'] = (df['Operating_Income'] / df['Revenue']) * 100

if {'Revenue', 'Net_Income'}.issubset(df.columns):
    df['Net_Profit_Margin'] = (df['Net_Income'] / df['Revenue']) * 100

if {'Revenue', 'EBITDA'}.issubset(df.columns):
    df['EBITDA_Margin'] = (df['EBITDA'] / df['Revenue']) * 100
if {'Net_Income', 'Total_Assets'}.issubset(df.columns):
    df['ROA'] = (df['Net_Income'] / df['Total_Assets']) * 100

if {'Net_Income', 'Total_Equity'}.issubset(df.columns):
    df['ROE'] = (df['Net_Income'] / df['Total_Equity']) * 100
if {'Total_Debt', 'Total_Assets'}.issubset(df.columns):
    df['Debt_to_Assets'] = df['Total_Debt'] / df['Total_Assets']

if {'Total_Debt', 'Total_Equity'}.issubset(df.columns):
    df['Debt_to_Equity'] = df['Total_Debt'] / df['Total_Equity']
if {'Current_Assets', 'Current_Liabilities'}.issubset(df.columns):
    df['Current_Ratio'] = (
        df['Current_Assets'] /
        df['Current_Liabilities']
    )

if {'Cash', 'Current_Liabilities'}.issubset(df.columns):
    df['Cash_Ratio'] = (
        df['Cash'] /
        df['Current_Liabilities']
    )
if {'Revenue', 'Total_Assets'}.issubset(df.columns):
    df['Asset_Turnover'] = (
        df['Revenue'] /
        df['Total_Assets']
    )
if {'Net_Income', 'Shares_Outstanding'}.issubset(df.columns):
    df['Calculated_EPS'] = (
        df['Net_Income'] /
        df['Shares_Outstanding']
    )
score_columns = [
    'Gross_Margin',
    'Operating_Margin',
    'Net_Profit_Margin',
    'ROA',
    'ROE'
]

available = [c for c in score_columns if c in df.columns]

if len(available) > 0:

    df['Financial_Health_Score'] = (
        df[available]
        .fillna(0)
        .mean(axis=1)
    )
if 'Financial_Health_Score' in df.columns:

    df['Company_Category'] = pd.cut(
        df['Financial_Health_Score'],
        bins=[-100, 0, 10, 20, 100],
        labels=[
            'Poor',
            'Average',
            'Good',
            'Excellent'
        ]
    )
if 'Revenue' in df.columns:

    df['Revenue_Category'] = pd.qcut(
        df['Revenue'],
        q=4,
        labels=[
            'Low',
            'Medium',
            'High',
            'Very High'
        ]
    )
if 'Debt_to_Equity' in df.columns:

    df['Debt_Level'] = pd.cut(
        df['Debt_to_Equity'],
        bins=[-100, 0.5, 1, 2, 100],
        labels=[
            'Low',
            'Moderate',
            'High',
            'Very High'
        ]
    )
df.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)

numeric_cols = df.select_dtypes(include=np.number).columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)
df.to_csv(
    "Financialdata_Feature_Engineered.csv",
    index=False
)

print("="*60)
print("Feature Engineering Completed Successfully")
print("="*60)

print("\nNew Shape:", df.shape)

print("\nNew Features Created:")

new_features = [
    'Gross_Margin',
    'Operating_Margin',
    'Net_Profit_Margin',
    'EBITDA_Margin',
    'ROA',
    'ROE',
    'Debt_to_Assets',
    'Debt_to_Equity',
    'Current_Ratio',
    'Cash_Ratio',
    'Asset_Turnover',
    'Calculated_EPS',
    'Financial_Health_Score',
    'Company_Category',
    'Revenue_Category',
    'Debt_Level'
]

for feature in new_features:
    if feature in df.columns:
        print("✔", feature)
```
